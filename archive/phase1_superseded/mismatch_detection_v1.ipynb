```python
# =============================================================================
# BOOK-TO-MOVIE MISMATCH DETECTION PIPELINE
# Input:  book_movie_pairs.csv (2392 pairs, pre-matched)
# Output: mismatch_scores.csv + EDA plots
#
# Columns available: book_title, movie_title, book_summary, movie_plot,
#                    book_genres, movie_genres, match_score
#
# Key design constraint:
#   book_summary mean=3128 chars, movie_plot mean=311 chars (10:1 ratio).
#   Raw comparison would measure LENGTH DIFFERENCE, not content divergence.
#   Every scoring head must account for this asymmetry explicitly.
#
# Four scoring heads:
#   A) Semantic distance   — embedding cosine dist on length-equalized text
#   B) Genre shift         — JSD between genre probability vectors (your RoBERTa)
#   C) Sentiment delta     — VADER compound score difference
#   D) Entity overlap      — spaCy NER, overlap on truncated book side
#
# Runtime estimate (Colab/Kaggle T4): ~55-70 minutes total
# =============================================================================

# ─────────────────────────────────────────────────────────────────────────────
# CELL 1 — INSTALL
# ─────────────────────────────────────────────────────────────────────────────
!pip install -q sentence-transformers vaderSentiment spacy
!python -m spacy download en_core_web_sm -q

from google.colab import drive
drive.mount('/content/drive')

import torch
print(f"PyTorch: {torch.__version__}")
print(f"GPU: {torch.cuda.get_device_name(0) if torch.cuda.is_available() else 'CPU'}")


# ─────────────────────────────────────────────────────────────────────────────
# CELL 2 — CONFIG & LOAD
# ─────────────────────────────────────────────────────────────────────────────
import pandas as pd
import numpy as np
import ast
import time
import pickle
from datetime import timedelta

DATA  = "/content/drive/MyDrive/nlp-genre-project-data/"
SAVE  = "/content/"

def fmt(s): return str(timedelta(seconds=int(s)))

df = pd.read_csv(f"{DATA}book_movie_pairs.csv")
print(f"Loaded {len(df)} pairs")
print(f"Book summary  — mean: {df['book_summary'].str.len().mean():.0f} chars, "
      f"median: {df['book_summary'].str.len().median():.0f}")
print(f"Movie plot    — mean: {df['movie_plot'].str.len().mean():.0f} chars, "
      f"median: {df['movie_plot'].str.len().median():.0f}")

# Parse genre strings (stored as string repr of lists)
def parse_genres(g):
    try:
        return ast.literal_eval(g)
    except:
        return []

df['book_genres_list']  = df['book_genres'].apply(parse_genres)
df['movie_genres_list'] = df['movie_genres'].apply(parse_genres)

# Quality filter: drop pairs where either text is extremely short
# (< 50 chars movie or < 100 chars book → uninformative)
n_before = len(df)
df = df[df['book_summary'].str.len() >= 100].copy()
df = df[df['movie_plot'].str.len()   >= 50].copy()
df = df.reset_index(drop=True)
print(f"After length filter: {len(df)} pairs (dropped {n_before - len(df)})")

TARGET_GENRES = ['Action','Comedy','Drama','Horror','Thriller',
                 'Sci-Fi','Fantasy','Romance','Animation',
                 'Documentary','Literary Fiction','Non-Fiction']
NUM_GENRES = len(TARGET_GENRES)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")


# ─────────────────────────────────────────────────────────────────────────────
# CELL 3 — TEXT PREPROCESSING
#
# Length asymmetry fix: truncate book_summary to MOVIE_CHAR_LIMIT chars
# for all computations that are sensitive to length (entity overlap,
# sentiment). Embedding uses full text on both sides but we note the caveat.
#
# Strategy: take the LAST N chars of book summary (not first N) because
# book summaries often front-load setup; the climax/resolution (most
# comparable to a movie plot) is at the end.
# ─────────────────────────────────────────────────────────────────────────────
MOVIE_CHAR_LIMIT = 500   # truncate book to this for length-sensitive comparisons
                          # ~median movie plot length (268 chars) + buffer

df['book_truncated'] = df['book_summary'].apply(
    lambda x: x[-MOVIE_CHAR_LIMIT:] if len(x) > MOVIE_CHAR_LIMIT else x
)

print(f"Book truncated length — mean: {df['book_truncated'].str.len().mean():.0f}, "
      f"median: {df['book_truncated'].str.len().median():.0f}")
print(f"Movie plot length    — mean: {df['movie_plot'].str.len().mean():.0f}, "
      f"median: {df['movie_plot'].str.len().median():.0f}")
print("Lengths now comparable for entity/sentiment scoring.")


# ─────────────────────────────────────────────────────────────────────────────
# CELL 4 — HEAD A: SEMANTIC DISTANCE
#
# Embed FULL book summary and FULL movie plot separately.
# Use cosine distance between the pair embeddings.
#
# Why full text here (not truncated): sentence-transformers pool across
# all tokens into one 384-dim vector, so length difference doesn't bias
# the embedding magnitude — only the content influences the direction.
# Cosine distance is magnitude-invariant.
#
# Runtime: ~15-20 min for 2×2392 texts on T4
# ─────────────────────────────────────────────────────────────────────────────
from sentence_transformers import SentenceTransformer

print("\n" + "="*60)
print("HEAD A — SEMANTIC DISTANCE")
print("="*60)

embedder = SentenceTransformer('all-MiniLM-L6-v2')
embedder = embedder.to(device)

t0 = time.time()
print("Embedding book summaries (full text)...")
book_embs = embedder.encode(
    df['book_summary'].tolist(),
    batch_size=64,
    show_progress_bar=True,
    convert_to_numpy=True,
    normalize_embeddings=True   # unit norm → dot product = cosine similarity
)

print("Embedding movie plots (full text)...")
movie_embs = embedder.encode(
    df['movie_plot'].tolist(),
    batch_size=64,
    show_progress_bar=True,
    convert_to_numpy=True,
    normalize_embeddings=True
)

print(f"Embedding done in {fmt(time.time()-t0)}")

# Cosine similarity → distance (unit norms so dot product = cosine sim)
cosine_sims      = (book_embs * movie_embs).sum(axis=1)   # (N,)
semantic_distance = 1.0 - cosine_sims                       # [0, 2] but practically [0, 1]

df['semantic_distance'] = semantic_distance.astype(np.float32)
df['cosine_similarity']  = cosine_sims.astype(np.float32)

print(f"Semantic distance stats:")
print(f"  Mean:   {df['semantic_distance'].mean():.4f}")
print(f"  Std:    {df['semantic_distance'].std():.4f}")
print(f"  Min:    {df['semantic_distance'].min():.4f}  (most similar)")
print(f"  Max:    {df['semantic_distance'].max():.4f}  (most different)")

# Save embeddings for reuse — avoid re-running if session crashes
np.save(f"{SAVE}book_embs.npy",  book_embs)
np.save(f"{SAVE}movie_embs.npy", movie_embs)
print("Embeddings saved.")

# Free GPU memory before next head
del embedder
torch.cuda.empty_cache()


# ─────────────────────────────────────────────────────────────────────────────
# CELL 5 — HEAD B: GENRE SHIFT
#
# Run your saved fine-tuned RoBERTa on both book_summary and movie_plot.
# Output: 12-dim sigmoid probability vector per text.
# Mismatch = Jensen-Shannon Divergence between the two genre distributions.
#
# JSD is symmetric and bounded [0, 1] in log-base-2. High JSD → genres
# predicted for book and movie are very different.
#
# Important: we run RoBERTa on the FULL text (truncated to 512 tokens
# internally), not the char-truncated version. Genre is a global property.
#
# Runtime: ~25-30 min for 2×2392 texts on T4
# ─────────────────────────────────────────────────────────────────────────────
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from transformers import AutoTokenizer, AutoModel

print("\n" + "="*60)
print("HEAD B — GENRE SHIFT (RoBERTa inference)")
print("="*60)

MODEL_NAME = "roberta-base"
MODEL_PATH = f"{DATA}roberta_optimal_best.pt"   # your saved baseline

class InferenceDataset(Dataset):
    def __init__(self, texts, tokenizer, max_len=512):
        enc = tokenizer(list(texts), max_length=max_len, padding="max_length",
                        truncation=True, return_tensors="pt")
        self.input_ids      = enc["input_ids"]
        self.attention_mask = enc["attention_mask"]
    def __len__(self): return len(self.input_ids)
    def __getitem__(self, i):
        return {"input_ids": self.input_ids[i], "attention_mask": self.attention_mask[i]}

class RoBERTaClassifier(nn.Module):
    def __init__(self, num_labels=12):
        super().__init__()
        self.roberta = AutoModel.from_pretrained(MODEL_NAME)
        self.drop    = nn.Dropout(0.3)
        self.clf     = nn.Linear(768, num_labels)
    def forward(self, input_ids, attention_mask):
        out = self.roberta(input_ids=input_ids, attention_mask=attention_mask)
        return self.clf(self.drop(out.last_hidden_state[:, 0, :]))

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

# Load model
roberta = RoBERTaClassifier(num_labels=NUM_GENRES).to(device)
state = torch.load(MODEL_PATH, map_location=device, weights_only=True)
roberta.load_state_dict(state, strict=False)
roberta.eval()
print(f"Loaded RoBERTa from {MODEL_PATH}")

def get_genre_probs(texts, model, tokenizer, batch_size=64, device=device):
    """Returns sigmoid probabilities, shape (N, 12)"""
    ds     = InferenceDataset(texts, tokenizer)
    loader = DataLoader(ds, batch_size=batch_size, shuffle=False, num_workers=0)
    all_probs = []
    with torch.no_grad():
        for batch in loader:
            logits = model(batch["input_ids"].to(device),
                          batch["attention_mask"].to(device))
            probs  = torch.sigmoid(logits).cpu().numpy()
            all_probs.append(probs)
    return np.vstack(all_probs)

t0 = time.time()
print("Running RoBERTa on book summaries...")
book_genre_probs  = get_genre_probs(df['book_summary'].tolist(), roberta, tokenizer)
print(f"  Done in {fmt(time.time()-t0)}")

t1 = time.time()
print("Running RoBERTa on movie plots...")
movie_genre_probs = get_genre_probs(df['movie_plot'].tolist(), roberta, tokenizer)
print(f"  Done in {fmt(time.time()-t1)}")
print(f"Total genre inference: {fmt(time.time()-t0)}")

# Save raw probs for analysis
np.save(f"{SAVE}book_genre_probs.npy",  book_genre_probs)
np.save(f"{SAVE}movie_genre_probs.npy", movie_genre_probs)

# Jensen-Shannon Divergence per pair
# Add epsilon to avoid log(0); normalize to sum=1 (sigmoid outputs don't sum to 1)
EPS = 1e-8
def row_normalize(P):
    P = np.clip(P, EPS, 1.0)
    return P / P.sum(axis=1, keepdims=True)

book_probs_norm  = row_normalize(book_genre_probs)
movie_probs_norm = row_normalize(movie_genre_probs)

def jsd_row(p, q):
    """Jensen-Shannon divergence per row pair, log base 2, returns [0,1]"""
    m = 0.5 * (p + q)
    def kl(a, b): return np.sum(a * np.log2(np.clip(a/b, EPS, None)), axis=1)
    return 0.5 * kl(p, m) + 0.5 * kl(q, m)

genre_shift = jsd_row(book_probs_norm, movie_probs_norm)
df['genre_shift_jsd'] = genre_shift.astype(np.float32)

print(f"\nGenre shift (JSD) stats:")
print(f"  Mean:  {df['genre_shift_jsd'].mean():.4f}")
print(f"  Std:   {df['genre_shift_jsd'].std():.4f}")
print(f"  Min:   {df['genre_shift_jsd'].min():.4f}")
print(f"  Max:   {df['genre_shift_jsd'].max():.4f}")

# Store per-genre predicted labels for output report
THRESHOLD = 0.5   # simple threshold for binary genre labels in report
df['book_genres_predicted']  = [
    [TARGET_GENRES[i] for i in range(NUM_GENRES) if book_genre_probs[j,i] > THRESHOLD]
    for j in range(len(df))
]
df['movie_genres_predicted'] = [
    [TARGET_GENRES[i] for i in range(NUM_GENRES) if movie_genre_probs[j,i] > THRESHOLD]
    for j in range(len(df))
]

# Genres added/dropped in movie vs book
df['genres_added']   = df.apply(
    lambda r: [g for g in r['movie_genres_predicted'] if g not in r['book_genres_predicted']], axis=1)
df['genres_dropped'] = df.apply(
    lambda r: [g for g in r['book_genres_predicted'] if g not in r['movie_genres_predicted']], axis=1)

del roberta
torch.cuda.empty_cache()


# ─────────────────────────────────────────────────────────────────────────────
# CELL 6 — HEAD C: SENTIMENT DELTA
#
# VADER (Valence Aware Dictionary and sEntiment Reasoner):
#   - Rule-based, no GPU needed, runs in seconds
#   - Returns compound score in [-1, 1]: -1=very negative, +1=very positive
#   - Sentiment delta = |compound_book_truncated - compound_movie|
#
# We use book_TRUNCATED here (not full) because VADER compound on a
# 3000-char text averages out to near-neutral regardless of tone —
# long texts regress to mean. Truncated last-500-chars captures the
# ending/resolution tone, which is the most adaptation-relevant signal.
# ─────────────────────────────────────────────────────────────────────────────
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer

print("\n" + "="*60)
print("HEAD C — SENTIMENT DELTA (VADER)")
print("="*60)

analyzer = SentimentIntensityAnalyzer()

t0 = time.time()
book_sentiment  = [analyzer.polarity_scores(t)['compound'] for t in df['book_truncated']]
movie_sentiment = [analyzer.polarity_scores(t)['compound'] for t in df['movie_plot']]
print(f"Sentiment scoring done in {fmt(time.time()-t0)}")

df['book_sentiment']     = book_sentiment
df['movie_sentiment']    = movie_sentiment
df['sentiment_delta']    = np.abs(
    np.array(book_sentiment) - np.array(movie_sentiment)
).astype(np.float32)
df['sentiment_direction'] = np.array(movie_sentiment) - np.array(book_sentiment)
# positive = movie is more positive than book (lightening)
# negative = movie is darker than book (darkening)

print(f"Sentiment delta stats:")
print(f"  Mean:  {df['sentiment_delta'].mean():.4f}")
print(f"  Std:   {df['sentiment_delta'].std():.4f}")
print(f"  Pairs where movie more positive (lightening): "
      f"{(df['sentiment_direction'] > 0.2).sum()}")
print(f"  Pairs where movie more negative (darkening):  "
      f"{(df['sentiment_direction'] < -0.2).sum()}")


# ─────────────────────────────────────────────────────────────────────────────
# CELL 7 — HEAD D: ENTITY OVERLAP
#
# spaCy NER extracts named entities (PERSON, ORG, GPE, EVENT, WORK_OF_ART).
# We compare entities in book_TRUNCATED vs movie_plot.
#
# entity_overlap_score = 1 - (|intersection| / |union|)
# High score = few shared entities = plot divergence signal.
#
# Using truncated book to match length. Named entities in last ~500 chars
# of book summary = characters/places present in the climax/resolution,
# which is what the movie plot most likely covers.
#
# Runtime: ~8-10 min for 2×2392 texts
# ─────────────────────────────────────────────────────────────────────────────
import spacy

print("\n" + "="*60)
print("HEAD D — ENTITY OVERLAP (spaCy NER)")
print("="*60)

nlp = spacy.load("en_core_web_sm", disable=["parser", "tagger", "lemmatizer"])
ENTITY_TYPES = {"PERSON", "GPE", "ORG", "EVENT", "WORK_OF_ART", "LOC"}

def extract_entities(text):
    doc   = nlp(text[:1000])   # spaCy limit safety
    ents  = {e.text.lower().strip() for e in doc.ents if e.label_ in ENTITY_TYPES}
    return ents

def entity_overlap(book_ents, movie_ents):
    """1 - Jaccard similarity. Higher = less overlap = more divergent."""
    if not book_ents and not movie_ents:
        return 0.5   # both empty → no signal, assign neutral
    union = book_ents | movie_ents
    inter = book_ents & movie_ents
    return 1.0 - len(inter) / len(union)

t0 = time.time()
book_ents_list  = [extract_entities(t) for t in df['book_truncated']]
movie_ents_list = [extract_entities(t) for t in df['movie_plot']]
print(f"NER done in {fmt(time.time()-t0)}")

df['entity_overlap_score'] = [
    entity_overlap(b, m) for b, m in zip(book_ents_list, movie_ents_list)
]
df['book_entities']  = [list(e) for e in book_ents_list]
df['movie_entities'] = [list(e) for e in movie_ents_list]
df['entities_lost']  = [
    list(b - m) for b, m in zip(book_ents_list, movie_ents_list)
]

print(f"Entity overlap stats (1=no shared entities, 0=identical):")
print(f"  Mean:  {df['entity_overlap_score'].mean():.4f}")
print(f"  Std:   {df['entity_overlap_score'].std():.4f}")


# ─────────────────────────────────────────────────────────────────────────────
# CELL 8 — COMPOSITE MISMATCH SCORE
#
# Weighted combination of the four heads.
# Weights reflect signal reliability given this dataset's constraints:
#
#   semantic_distance  0.40  — most reliable; magnitude-invariant, full text
#   genre_shift_jsd    0.35  — domain-specific strength (your trained model)
#   entity_overlap     0.15  — noisy (short movie texts have few entities)
#   sentiment_delta    0.10  — weakest; VADER struggles with narrative text
#
# All four heads are min-max normalized to [0,1] before combining so that
# the different scales (JSD vs cosine dist vs absolute sentiment diff)
# don't dominate by unit accident.
# ─────────────────────────────────────────────────────────────────────────────
print("\n" + "="*60)
print("COMPOSITE MISMATCH SCORE")
print("="*60)

WEIGHTS = {
    'semantic_distance':  0.40,
    'genre_shift_jsd':    0.35,
    'entity_overlap_score': 0.15,
    'sentiment_delta':    0.10,
}

def min_max_normalize(series):
    mn, mx = series.min(), series.max()
    if mx == mn:
        return series * 0.0
    return (series - mn) / (mx - mn)

# Normalize each head to [0,1]
for col in WEIGHTS:
    df[f'{col}_norm'] = min_max_normalize(df[col])

# Weighted composite
df['composite_mismatch'] = sum(
    WEIGHTS[col] * df[f'{col}_norm'] for col in WEIGHTS
)

# Categorize into three tiers
# Thresholds chosen at 33rd and 67th percentile — data-driven, not arbitrary
p33 = df['composite_mismatch'].quantile(0.33)
p67 = df['composite_mismatch'].quantile(0.67)

def categorize(score):
    if score <= p33: return "Faithful"
    if score <= p67: return "Moderate"
    return "Significant"

df['mismatch_category'] = df['composite_mismatch'].apply(categorize)

print(f"Composite score stats:")
print(f"  Mean:   {df['composite_mismatch'].mean():.4f}")
print(f"  Std:    {df['composite_mismatch'].std():.4f}")
print(f"  p33:    {p33:.4f} (Faithful threshold)")
print(f"  p67:    {p67:.4f} (Significant threshold)")
print(f"\nMismatch category distribution:")
print(df['mismatch_category'].value_counts())


# ─────────────────────────────────────────────────────────────────────────────
# CELL 9 — OUTPUT
# ─────────────────────────────────────────────────────────────────────────────
print("\n" + "="*60)
print("SAVING OUTPUT")
print("="*60)

output_cols = [
    'book_title', 'movie_title', 'match_score',
    'composite_mismatch', 'mismatch_category',
    'semantic_distance', 'cosine_similarity',
    'genre_shift_jsd',
    'sentiment_delta', 'book_sentiment', 'movie_sentiment', 'sentiment_direction',
    'entity_overlap_score',
    'book_genres_list', 'movie_genres_list',
    'book_genres_predicted', 'movie_genres_predicted',
    'genres_added', 'genres_dropped',
    'book_entities', 'movie_entities', 'entities_lost',
]

df_out = df[output_cols].sort_values('composite_mismatch', ascending=False)
df_out.to_csv(f"{SAVE}mismatch_scores.csv", index=False)
df_out.to_csv(f"{DATA}mismatch_scores.csv", index=False)   # Drive backup
print(f"Saved mismatch_scores.csv ({len(df_out)} rows)")

# Top 15 most faithful and most diverged for sanity check
print("\n── TOP 15 MOST FAITHFUL (lowest mismatch) ──────────────────")
faithful = df_out.tail(15)[['book_title','movie_title','composite_mismatch',
                             'semantic_distance','genre_shift_jsd']]
print(faithful.to_string(index=False))

print("\n── TOP 15 MOST DIVERGED (highest mismatch) ─────────────────")
diverged = df_out.head(15)[['book_title','movie_title','composite_mismatch',
                             'semantic_distance','genre_shift_jsd']]
print(diverged.to_string(index=False))


# ─────────────────────────────────────────────────────────────────────────────
# CELL 10 — EDA & VISUALIZATION
# ─────────────────────────────────────────────────────────────────────────────
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
import seaborn as sns

print("\n" + "="*60)
print("EDA PLOTS")
print("="*60)

fig = plt.figure(figsize=(20, 22))
gs  = gridspec.GridSpec(4, 2, figure=fig, hspace=0.45, wspace=0.35)

colors = {'Faithful': '#2ecc71', 'Moderate': '#f39c12', 'Significant': '#e74c3c'}
cat_order = ['Faithful', 'Moderate', 'Significant']

# ── Plot 1: Composite score distribution ─────────────────────────────────────
ax1 = fig.add_subplot(gs[0, 0])
for cat in cat_order:
    sub = df_out[df_out['mismatch_category'] == cat]['composite_mismatch']
    ax1.hist(sub, bins=30, alpha=0.7, label=cat, color=colors[cat])
ax1.axvline(p33, color='gray', linestyle='--', linewidth=1, label=f'p33={p33:.2f}')
ax1.axvline(p67, color='gray', linestyle='-',  linewidth=1, label=f'p67={p67:.2f}')
ax1.set_title('Composite Mismatch Score Distribution', fontsize=13, fontweight='bold')
ax1.set_xlabel('Composite Mismatch Score')
ax1.set_ylabel('Count')
ax1.legend()

# ── Plot 2: Head correlation heatmap ─────────────────────────────────────────
ax2 = fig.add_subplot(gs[0, 1])
corr_cols = ['composite_mismatch', 'semantic_distance',
             'genre_shift_jsd', 'sentiment_delta', 'entity_overlap_score']
corr = df_out[corr_cols].corr()
sns.heatmap(corr, annot=True, fmt='.2f', cmap='RdYlGn_r',
            ax=ax2, square=True, linewidths=0.5,
            xticklabels=['Composite','Semantic','Genre','Sentiment','Entity'],
            yticklabels=['Composite','Semantic','Genre','Sentiment','Entity'])
ax2.set_title('Head Correlation Matrix\n(helps assess redundancy)', fontsize=13, fontweight='bold')

# ── Plot 3: Semantic distance vs Genre shift scatter ─────────────────────────
ax3 = fig.add_subplot(gs[1, 0])
for cat in cat_order:
    sub = df_out[df_out['mismatch_category'] == cat]
    ax3.scatter(sub['semantic_distance'], sub['genre_shift_jsd'],
                alpha=0.4, s=12, label=cat, color=colors[cat])
ax3.set_xlabel('Semantic Distance (embedding cosine dist)')
ax3.set_ylabel('Genre Shift (JSD)')
ax3.set_title('Semantic Distance vs Genre Shift\n(main two heads)', fontsize=13, fontweight='bold')
ax3.legend()

# ── Plot 4: Sentiment direction distribution ──────────────────────────────────
ax4 = fig.add_subplot(gs[1, 1])
ax4.hist(df_out['sentiment_direction'], bins=40, color='steelblue', alpha=0.8, edgecolor='white')
ax4.axvline(0, color='red', linestyle='--', linewidth=1.5, label='No change')
ax4.axvline(df_out['sentiment_direction'].mean(), color='orange', linestyle='-',
            linewidth=1.5, label=f"Mean={df_out['sentiment_direction'].mean():.3f}")
ax4.set_title('Sentiment Direction\n(positive = movie lighter than book)', fontsize=13, fontweight='bold')
ax4.set_xlabel('Movie Sentiment − Book Sentiment')
ax4.set_ylabel('Count')
ax4.legend()

# ── Plot 5: Genre shift by mismatch category (boxplot) ───────────────────────
ax5 = fig.add_subplot(gs[2, 0])
data_by_cat = [df_out[df_out['mismatch_category']==c]['genre_shift_jsd'].values
               for c in cat_order]
bp = ax5.boxplot(data_by_cat, labels=cat_order, patch_artist=True,
                 medianprops=dict(color='black', linewidth=2))
for patch, cat in zip(bp['boxes'], cat_order):
    patch.set_facecolor(colors[cat])
    patch.set_alpha(0.7)
ax5.set_title('Genre Shift by Mismatch Category', fontsize=13, fontweight='bold')
ax5.set_ylabel('Genre Shift (JSD)')

# ── Plot 6: Genre transition heatmap ─────────────────────────────────────────
# For each book genre, what movie genres appear most often?
ax6 = fig.add_subplot(gs[2, 1])
genre_transition = np.zeros((NUM_GENRES, NUM_GENRES))
for _, row in df_out.iterrows():
    for bg in row['book_genres_list']:
        if bg in TARGET_GENRES:
            bi = TARGET_GENRES.index(bg)
            for mg in row['movie_genres_list']:
                if mg in TARGET_GENRES:
                    mi = TARGET_GENRES.index(mg)
                    genre_transition[bi, mi] += 1

# Row-normalize (conditional probability P(movie_genre | book_genre))
row_sums = genre_transition.sum(axis=1, keepdims=True)
row_sums = np.where(row_sums == 0, 1, row_sums)
genre_transition_norm = genre_transition / row_sums

short_names = ['Act','Com','Dra','Hor','Thr','Sci','Fan','Rom','Ani','Doc','Lit','Non']
sns.heatmap(genre_transition_norm, annot=True, fmt='.2f', cmap='Blues',
            ax=ax6, xticklabels=short_names, yticklabels=short_names,
            linewidths=0.3, vmin=0, vmax=0.5)
ax6.set_title('Genre Transition Matrix\nP(movie genre | book genre)', fontsize=13, fontweight='bold')
ax6.set_xlabel('Movie Genre')
ax6.set_ylabel('Book Genre')

# ── Plot 7: Top 20 most & least faithful pairs ────────────────────────────────
ax7 = fig.add_subplot(gs[3, :])
top_n = 10
most_faithful  = df_out.nsmallest(top_n, 'composite_mismatch')
most_diverged  = df_out.nlargest(top_n, 'composite_mismatch')
examples       = pd.concat([most_diverged, most_faithful])
labels         = [f"{r['book_title'][:25]} → {r['movie_title'][:25]}"
                  for _, r in examples.iterrows()]
scores         = examples['composite_mismatch'].values
bar_colors     = ['#e74c3c']*top_n + ['#2ecc71']*top_n

bars = ax7.barh(range(len(labels)), scores, color=bar_colors, alpha=0.8)
ax7.set_yticks(range(len(labels)))
ax7.set_yticklabels(labels, fontsize=8)
ax7.axvline(p33, color='gray', linestyle='--', linewidth=1)
ax7.axvline(p67, color='gray', linestyle='-',  linewidth=1)
ax7.set_title('Top 10 Most Diverged (red) & Most Faithful (green) Adaptations',
              fontsize=13, fontweight='bold')
ax7.set_xlabel('Composite Mismatch Score')
ax7.invert_yaxis()

plt.suptitle('Book-to-Movie Mismatch Analysis\n2392 Paired Book/Film Summaries',
             fontsize=16, fontweight='bold', y=1.01)

plt.savefig(f"{SAVE}mismatch_analysis.png", dpi=150, bbox_inches='tight')
plt.savefig(f"{DATA}mismatch_analysis.png", dpi=150, bbox_inches='tight')
plt.show()
print("Plots saved.")


# ─────────────────────────────────────────────────────────────────────────────
# CELL 11 — SANITY CHECK ON KNOWN ADAPTATIONS
#
# Validate the metric against adaptations with known fidelity.
# If your metric ranks these correctly, it's defensible to your professor.
# ─────────────────────────────────────────────────────────────────────────────
print("\n" + "="*60)
print("SANITY CHECK — KNOWN ADAPTATIONS")
print("="*60)

# Known faithful adaptations (should score LOW mismatch)
faithful_known = [
    'no country for old men', 'the godfather', 'harry potter',
    'fight club', 'silence of the lambs', 'schindler'
]
# Known divergent adaptations (should score HIGH mismatch)
divergent_known = [
    'world war z', 'the shining', 'i am legend',
    'forrest gump', 'cloud atlas', 'american psycho'
]

print("\nSearching for known adaptations in dataset...")
df_lower = df_out.copy()
df_lower['book_lower']  = df_lower['book_title'].str.lower()
df_lower['movie_lower'] = df_lower['movie_title'].str.lower()

print("\n-- Expected FAITHFUL (should have LOW mismatch score) --")
for name in faithful_known:
    mask = (df_lower['book_lower'].str.contains(name, na=False) |
            df_lower['movie_lower'].str.contains(name, na=False))
    if mask.any():
        row = df_lower[mask].iloc[0]
        print(f"  {row['book_title'][:30]:<32} score={row['composite_mismatch']:.3f}  "
              f"[{row['mismatch_category']}]")
    else:
        print(f"  '{name}' — NOT IN DATASET")

print("\n-- Expected DIVERGENT (should have HIGH mismatch score) --")
for name in divergent_known:
    mask = (df_lower['book_lower'].str.contains(name, na=False) |
            df_lower['movie_lower'].str.contains(name, na=False))
    if mask.any():
        row = df_lower[mask].iloc[0]
        print(f"  {row['book_title'][:30]:<32} score={row['composite_mismatch']:.3f}  "
              f"[{row['mismatch_category']}]")
    else:
        print(f"  '{name}' — NOT IN DATASET")

print("\n✅ Pipeline complete.")
print(f"Output files:")
print(f"  {SAVE}mismatch_scores.csv")
print(f"  {SAVE}mismatch_analysis.png")
print(f"  {DATA}mismatch_scores.csv  (Drive backup)")
print(f"  {DATA}mismatch_analysis.png (Drive backup)")
```
