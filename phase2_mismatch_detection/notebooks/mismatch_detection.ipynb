```python
# =============================================================================
# BOOK-TO-MOVIE MISMATCH DETECTION — FINAL VERSION
# Input:  wiki_book_movie_pairs.csv
# Output: wiki_mismatch_scores.csv + wiki_mismatch_analysis.png
#
# FIXES vs previous version:
#   1. WRONG-MATCH FILTER: pairs where semantic cosine_sim < WRONG_MATCH_THRESHOLD
#      are removed BEFORE scoring; printed separately for review.
#      These are scraper errors (War and Peace matched to The Peace War etc.),
#      not genuine high-mismatch pairs. Without this filter they contaminate
#      the top of the diverged list with values near 1.0, making the rest of
#      the distribution meaningless.
#
#   2. GENRE TRANSITION MATRIX FIXED: now uses RoBERTa-predicted genres
#      (book_genres_predicted / movie_genres_predicted) instead of
#      movie_genres_list which is always [] because Wikipedia has no
#      structured genre metadata.
#
#   3. COMPOSITE WEIGHTS rebalanced: entity and TF-IDF correlate at 0.81
#      → entity weight reduced; semantic weight increased.
#      New weights: semantic 0.35 / genre 0.25 / tfidf 0.25 / entity 0.10 / sentiment 0.05
#
#   4. SANITY CHECK strengthened: now reports exact scores for all known pairs
#      and flags any ✅/❌ result with the head that is responsible for
#      unexpected direction, to aid diagnosis.
#
#   5. VADER note: sentiment_delta intentionally uses first 1500 chars only
#      (VADER regresses toward 0 on long texts). The mean delta of ~0.63
#      reflects register difference (literary vs encyclopedic prose), not
#      signal noise. Low weight (0.05) is appropriate.
#
# Runtime on Colab T4: ~40-50 min
# =============================================================================

# BOOK-TO-MOVIE MISMATCH DETECTION — FINAL VERSION
# Input:  wiki_book_movie_pairs.csv
# Output: wiki_mismatch_scores.csv + wiki_mismatch_analysis.png

# ─────────────────────────────────────────────────────────────────────────────
# CELL 1 — INSTALL & MOUNT
# ─────────────────────────────────────────────────────────────────────────────
!pip install -q sentence-transformers vaderSentiment spacy scikit-learn
!python -m spacy download en_core_web_sm -q

from google.colab import drive
drive.mount('/content/drive')

import torch
print(f"PyTorch: {torch.__version__}")
print(f"GPU: {torch.cuda.get_device_name(0) if torch.cuda.is_available() else 'CPU'}")


# ─────────────────────────────────────────────────────────────────────────────
# CELL 2 — CONFIG, LOAD & WRONG-MATCH FILTER
# ─────────────────────────────────────────────────────────────────────────────
import pandas as pd
import numpy as np
import ast, time, re
from datetime import timedelta
from collections import defaultdict

DATA = "/content/drive/MyDrive/nlp-genre-project-data/"
SAVE = "/content/"

def fmt(s): return str(timedelta(seconds=int(s)))

# ── Load ──────────────────────────────────────────────────────────────────────
df_raw = pd.read_csv(f"{DATA}wiki_book_movie_pairs.csv")
print(f"Loaded {len(df_raw)} pairs")

def parse_genres(g):
    try:
        result = ast.literal_eval(str(g))
        return result if isinstance(result, list) else []
    except:
        return []

df_raw["book_genres_list"]  = df_raw["book_genres"].apply(parse_genres)
df_raw["movie_genres_list"] = df_raw["movie_genres"].apply(parse_genres)

# Basic quality filter
df_raw = df_raw[df_raw["book_summary"].str.len() >= 100].copy()
df_raw = df_raw[df_raw["movie_plot"].str.len()   >= 200].copy()
df_raw = df_raw.reset_index(drop=True)
print(f"After quality filter: {len(df_raw)} pairs")

# ── Wrong-match filter via fast embedding cosine similarity ───────────────────
# We compute embeddings once here and reuse them in HEAD A.
# Pairs with cosine_sim < WRONG_MATCH_THRESHOLD are scraper errors:
# the book and film have essentially nothing in common semantically.
# These are NOT high-mismatch pairs — they are wrong pairs.
# They must be removed BEFORE scoring, not kept and scored near 1.0.
WRONG_MATCH_THRESHOLD = 0.20   # cosine similarity floor

from sentence_transformers import SentenceTransformer
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

embedder_tmp = SentenceTransformer("all-MiniLM-L6-v2").to(device)
print("Computing pre-filter embeddings for wrong-match detection...")
book_embs_raw = embedder_tmp.encode(
    df_raw["book_summary"].tolist(), batch_size=64,
    show_progress_bar=False, convert_to_numpy=True, normalize_embeddings=True)
movie_embs_raw = embedder_tmp.encode(
    df_raw["movie_plot"].tolist(), batch_size=64,
    show_progress_bar=False, convert_to_numpy=True, normalize_embeddings=True)
del embedder_tmp
torch.cuda.empty_cache()

cosine_raw = (book_embs_raw * movie_embs_raw).sum(axis=1)

# Identify wrong matches
wrong_mask = cosine_raw < WRONG_MATCH_THRESHOLD
df_wrong   = df_raw[wrong_mask].copy()
df_wrong["cosine_prefilter"] = cosine_raw[wrong_mask]
df         = df_raw[~wrong_mask].copy().reset_index(drop=True)
book_embs  = book_embs_raw[~wrong_mask]
movie_embs = movie_embs_raw[~wrong_mask]
del book_embs_raw, movie_embs_raw

print(f"\n{'='*60}")
print(f"WRONG-MATCH FILTER  (cosine_sim < {WRONG_MATCH_THRESHOLD})")
print(f"{'='*60}")
print(f"  Removed: {len(df_wrong)} pairs  |  Kept: {len(df)} pairs")
if len(df_wrong) > 0:
    print(f"\n  Removed pairs (scraper errors, NOT high-mismatch):")
    for _, row in df_wrong.sort_values("cosine_prefilter").iterrows():
        print(f"    sim={row['cosine_prefilter']:.3f}  "
              f"'{row['book_title'][:35]}' → '{row['movie_title'][:35]}'")
df_wrong.to_csv(f"{SAVE}wiki_wrong_matches.csv", index=False)
print(f"\n  Saved to wiki_wrong_matches.csv for reference.")

# ── Stats after filter ────────────────────────────────────────────────────────
book_lens  = df["book_summary"].str.len()
movie_lens = df["movie_plot"].str.len()
print(f"\nBook summary  — mean:{book_lens.mean():.0f}  median:{book_lens.median():.0f}  chars")
print(f"Movie plot    — mean:{movie_lens.mean():.0f}  median:{movie_lens.median():.0f}  chars")
print(f"Length ratio  — mean:{(book_lens/movie_lens).mean():.1f}x  (TMDB was 10x, target <3x)")

# Strategy confidence weights
STRATEGY_CONFIDENCE = {
    "pass1_wikidata_P144":       1.00,
    "pass1_wikidata_P144_batch": 1.00,
    "pass2_book_adaptations":    0.90,
    "pass3_author_category":     0.85,
    "pass4_fuzzy_verified":      0.70,
}
df["confidence"] = df["strategy"].map(STRATEGY_CONFIDENCE).fillna(0.80)

TARGET_GENRES = ["Action","Comedy","Drama","Horror","Thriller",
                 "Sci-Fi","Fantasy","Romance","Animation",
                 "Documentary","Literary Fiction","Non-Fiction"]
NUM_GENRES = len(TARGET_GENRES)


# ─────────────────────────────────────────────────────────────────────────────
# CELL 3 — HEAD A: SEMANTIC DISTANCE
# Embeddings already computed above — just measure cosine distance.
# ─────────────────────────────────────────────────────────────────────────────
print(f"\n{'='*55}\nHEAD A — SEMANTIC DISTANCE\n{'='*55}")

cosine_sims            = (book_embs * movie_embs).sum(axis=1)
df["cosine_similarity"]  = cosine_sims.astype(np.float32)
df["semantic_distance"]  = (1.0 - cosine_sims).astype(np.float32)

print(f"Semantic distance — mean:{df['semantic_distance'].mean():.4f}  "
      f"std:{df['semantic_distance'].std():.4f}  "
      f"range:[{df['semantic_distance'].min():.3f}, {df['semantic_distance'].max():.3f}]")
print(f"(All pairs have cosine_sim ≥ {WRONG_MATCH_THRESHOLD} — wrong matches removed)")

np.save(f"{SAVE}wiki_book_embs.npy",  book_embs)
np.save(f"{SAVE}wiki_movie_embs.npy", movie_embs)


# ─────────────────────────────────────────────────────────────────────────────
# CELL 4 — HEAD B: GENRE SHIFT (RoBERTa)
# ─────────────────────────────────────────────────────────────────────────────
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from transformers import AutoTokenizer, AutoModel

print(f"\n{'='*55}\nHEAD B — GENRE SHIFT (RoBERTa)\n{'='*55}")

MODEL_NAME = "roberta-base"
MODEL_PATH = f"{DATA}roberta_optimal_best.pt"

class InferenceDataset(Dataset):
    def __init__(self, texts, tokenizer, max_len=512):
        enc = tokenizer(list(texts), max_length=max_len, padding="max_length",
                        truncation=True, return_tensors="pt")
        self.input_ids      = enc["input_ids"]
        self.attention_mask = enc["attention_mask"]
    def __len__(self): return len(self.input_ids)
    def __getitem__(self, i):
        return {"input_ids":      self.input_ids[i],
                "attention_mask": self.attention_mask[i]}

class RoBERTaClassifier(nn.Module):
    def __init__(self, num_labels=12):
        super().__init__()
        self.roberta = AutoModel.from_pretrained(MODEL_NAME)
        self.drop    = nn.Dropout(0.3)
        self.clf     = nn.Linear(768, num_labels)
    def forward(self, input_ids, attention_mask):
        out = self.roberta(input_ids=input_ids, attention_mask=attention_mask)
        return self.clf(self.drop(out.last_hidden_state[:, 0, :]))

tokenizer  = AutoTokenizer.from_pretrained(MODEL_NAME)
roberta    = RoBERTaClassifier(NUM_GENRES).to(device)
state_dict = torch.load(MODEL_PATH, map_location=device, weights_only=True)
roberta.load_state_dict(state_dict, strict=False)
roberta.eval()
print(f"Loaded RoBERTa from {MODEL_PATH}")

def get_genre_probs(texts, model, tokenizer, batch_size=64):
    ds     = InferenceDataset(texts, tokenizer)
    loader = DataLoader(ds, batch_size=batch_size, shuffle=False, num_workers=0)
    probs  = []
    with torch.no_grad():
        for batch in loader:
            logits = model(batch["input_ids"].to(device),
                          batch["attention_mask"].to(device))
            probs.append(torch.sigmoid(logits).cpu().numpy())
    return np.vstack(probs)

t0 = time.time()
print("Running RoBERTa on book summaries...")
book_genre_probs  = get_genre_probs(df["book_summary"].tolist(), roberta, tokenizer)
print(f"  Done in {fmt(time.time()-t0)}")

t1 = time.time()
print("Running RoBERTa on movie plots...")
movie_genre_probs = get_genre_probs(df["movie_plot"].tolist(), roberta, tokenizer)
print(f"  Done in {fmt(time.time()-t1)}")

np.save(f"{SAVE}wiki_book_genre_probs.npy",  book_genre_probs)
np.save(f"{SAVE}wiki_movie_genre_probs.npy", movie_genre_probs)

# JSD between genre distributions
EPS = 1e-8
def row_normalize(P):
    P = np.clip(P, EPS, 1.0)
    return P / P.sum(axis=1, keepdims=True)

def jsd_rows(p, q):
    p, q = row_normalize(p), row_normalize(q)
    m    = 0.5 * (p + q)
    kl   = lambda a, b: np.sum(a * np.log2(np.clip(a/b, EPS, None)), axis=1)
    return np.clip(0.5 * kl(p, m) + 0.5 * kl(q, m), 0, 1)

df["genre_shift_jsd"] = jsd_rows(book_genre_probs, movie_genre_probs).astype(np.float32)
print(f"Genre shift JSD — mean:{df['genre_shift_jsd'].mean():.4f}  "
      f"std:{df['genre_shift_jsd'].std():.4f}")

# Genre predictions (used for genre transition heatmap — fixes all-zeros bug)
THRESHOLD = 0.5
df["book_genres_predicted"]  = [
    [TARGET_GENRES[i] for i in range(NUM_GENRES) if book_genre_probs[j,i]  > THRESHOLD]
    for j in range(len(df))]
df["movie_genres_predicted"] = [
    [TARGET_GENRES[i] for i in range(NUM_GENRES) if movie_genre_probs[j,i] > THRESHOLD]
    for j in range(len(df))]
df["genres_added"]   = df.apply(
    lambda r: [g for g in r["movie_genres_predicted"]
               if g not in r["book_genres_predicted"]], axis=1)
df["genres_dropped"] = df.apply(
    lambda r: [g for g in r["book_genres_predicted"]
               if g not in r["movie_genres_predicted"]], axis=1)

del roberta
torch.cuda.empty_cache()


# ─────────────────────────────────────────────────────────────────────────────
# CELL 5 — HEAD C: SENTIMENT DELTA (VADER)
# NOTE: VADER regresses toward 0 on long texts (averages out all signal).
# Use first 1500 chars of each text — captures opening tone, not resolution.
# Both texts are now similar length so the truncation is symmetric.
# Mean delta ~0.6 reflects register difference (literary prose vs encyclopedic)
# not pure noise. Low weight (0.05) is intentional.
# ─────────────────────────────────────────────────────────────────────────────
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer

print(f"\n{'='*55}\nHEAD C — SENTIMENT DELTA (VADER)\n{'='*55}")

VADER_CHARS = 1500
analyzer = SentimentIntensityAnalyzer()

t0 = time.time()
book_sent  = [analyzer.polarity_scores(t[:VADER_CHARS])["compound"]
              for t in df["book_summary"]]
movie_sent = [analyzer.polarity_scores(t[:VADER_CHARS])["compound"]
              for t in df["movie_plot"]]
print(f"Done in {fmt(time.time()-t0)}")

df["book_sentiment"]      = book_sent
df["movie_sentiment"]     = movie_sent
df["sentiment_delta"]     = np.abs(np.array(book_sent) - np.array(movie_sent)).astype(np.float32)
df["sentiment_direction"] = (np.array(movie_sent) - np.array(book_sent)).astype(np.float32)

print(f"Sentiment delta — mean:{df['sentiment_delta'].mean():.4f}  "
      f"(high due to prose register difference; low weight 0.05 applied)")
print(f"Movie lighter than book (>0.2): {(df['sentiment_direction'] > 0.2).sum()}")
print(f"Movie darker  than book (<-0.2): {(df['sentiment_direction'] < -0.2).sum()}")


# ─────────────────────────────────────────────────────────────────────────────
# CELL 6 — HEAD D: ENTITY OVERLAP (spaCy NER)
# ─────────────────────────────────────────────────────────────────────────────
import spacy

print(f"\n{'='*55}\nHEAD D — ENTITY OVERLAP (spaCy NER)\n{'='*55}")

nlp = spacy.load("en_core_web_sm", disable=["parser","tagger","lemmatizer"])
ENTITY_TYPES  = {"PERSON","GPE","ORG","EVENT","WORK_OF_ART","LOC"}
SPACY_CHAR_LIMIT = 5000

def extract_entities(text):
    doc = nlp(text[:SPACY_CHAR_LIMIT])
    return {e.text.lower().strip() for e in doc.ents if e.label_ in ENTITY_TYPES}

def jaccard_distance(a, b):
    if not a and not b: return 0.5
    return 1.0 - len(a & b) / len(a | b)

t0 = time.time()
book_ents  = [extract_entities(t) for t in df["book_summary"]]
movie_ents = [extract_entities(t) for t in df["movie_plot"]]
print(f"NER done in {fmt(time.time()-t0)}")

df["entity_overlap_score"] = [jaccard_distance(b, m)
                               for b, m in zip(book_ents, movie_ents)]
df["book_entities"]   = [list(e) for e in book_ents]
df["movie_entities"]  = [list(e) for e in movie_ents]
df["entities_lost"]   = [list(b - m) for b, m in zip(book_ents, movie_ents)]
df["entities_gained"] = [list(m - b) for b, m in zip(book_ents, movie_ents)]

print(f"Entity overlap — mean:{df['entity_overlap_score'].mean():.4f}  "
      f"(1=no overlap, 0=identical entities)")
print(f"Mean entities book:{np.mean([len(e) for e in book_ents]):.1f}  "
      f"movie:{np.mean([len(e) for e in movie_ents]):.1f}")


# ─────────────────────────────────────────────────────────────────────────────
# CELL 7 — HEAD E: NARRATIVE SIMILARITY (TF-IDF)
# ─────────────────────────────────────────────────────────────────────────────
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity as sk_cosine

print(f"\n{'='*55}\nHEAD E — NARRATIVE SIMILARITY (TF-IDF)\n{'='*55}")

all_texts  = df["book_summary"].tolist() + df["movie_plot"].tolist()
vectorizer = TfidfVectorizer(
    stop_words="english", max_features=15000,
    ngram_range=(1, 2), min_df=2, sublinear_tf=True
)
vectorizer.fit(all_texts)
book_tfidf  = vectorizer.transform(df["book_summary"].tolist())
movie_tfidf = vectorizer.transform(df["movie_plot"].tolist())

tfidf_sims = []
BATCH = 50
for i in range(0, len(df), BATCH):
    sims = sk_cosine(book_tfidf[i:i+BATCH], movie_tfidf[i:i+BATCH]).diagonal()
    tfidf_sims.extend(sims.tolist())

df["tfidf_similarity"] = np.array(tfidf_sims).astype(np.float32)
df["tfidf_distance"]   = (1.0 - df["tfidf_similarity"]).astype(np.float32)

print(f"TF-IDF distance — mean:{df['tfidf_distance'].mean():.4f}  "
      f"std:{df['tfidf_distance'].std():.4f}")
print(f"  (lower = more shared vocabulary = more faithful)")


# ─────────────────────────────────────────────────────────────────────────────
# CELL 8 — COMPOSITE MISMATCH SCORE
#
# Weight rationale:
#   semantic_distance  0.35 — global thematic distance (most reliable)
#   genre_shift_jsd    0.25 — genre-level mismatch (domain-specific)
#   tfidf_distance     0.25 — shared plot vocabulary
#   entity_overlap     0.10 — (reduced from 0.15: correlates 0.81 with TF-IDF)
#   sentiment_delta    0.05 — tone (low: high baseline delta from register diff)
#
# confidence multiplier: downweights Bronze matches in ranking
# ─────────────────────────────────────────────────────────────────────────────
print(f"\n{'='*55}\nCOMPOSITE MISMATCH SCORE\n{'='*55}")

WEIGHTS = {
    "semantic_distance":    0.35,
    "genre_shift_jsd":      0.25,
    "tfidf_distance":       0.25,
    "entity_overlap_score": 0.10,
    "sentiment_delta":      0.05,
}
assert abs(sum(WEIGHTS.values()) - 1.0) < 1e-9, "Weights must sum to 1.0"

def min_max_norm(s):
    mn, mx = s.min(), s.max()
    return (s - mn) / (mx - mn) if mx > mn else pd.Series(np.zeros(len(s)))

for col in WEIGHTS:
    df[f"{col}_norm"] = min_max_norm(df[col])

df["composite_raw"]      = sum(w * df[f"{col}_norm"] for col, w in WEIGHTS.items())
df["composite_mismatch"] = min_max_norm(df["composite_raw"] * df["confidence"])

p33 = df["composite_mismatch"].quantile(0.33)
p67 = df["composite_mismatch"].quantile(0.67)
df["mismatch_category"] = pd.cut(
    df["composite_mismatch"],
    bins=[-0.001, p33, p67, 1.001],
    labels=["Faithful","Moderate","Significant"]
)

print(f"Composite score — mean:{df['composite_mismatch'].mean():.4f}  "
      f"std:{df['composite_mismatch'].std():.4f}")
print(f"p33={p33:.3f}  p67={p67:.3f}")
print(f"\nCategory distribution:")
print(df["mismatch_category"].value_counts().to_string())


# ─────────────────────────────────────────────────────────────────────────────
# CELL 9 — SAVE OUTPUT
# ─────────────────────────────────────────────────────────────────────────────
output_cols = [
    "book_title","movie_title","strategy","confidence","match_score",
    "composite_mismatch","composite_raw","mismatch_category",
    "semantic_distance","cosine_similarity",
    "genre_shift_jsd",
    "tfidf_distance","tfidf_similarity",
    "sentiment_delta","book_sentiment","movie_sentiment","sentiment_direction",
    "entity_overlap_score",
    "book_genres_list","movie_genres_list",
    "book_genres_predicted","movie_genres_predicted",
    "genres_added","genres_dropped",
    "book_entities","movie_entities","entities_lost","entities_gained",
]
df_out = df[output_cols].sort_values("composite_mismatch", ascending=False)
df_out.to_csv(f"{SAVE}wiki_mismatch_scores.csv", index=False)
df_out.to_csv(f"{DATA}wiki_mismatch_scores.csv", index=False)
print(f"Saved wiki_mismatch_scores.csv ({len(df_out)} rows)")

print("\n── TOP 15 MOST FAITHFUL ─────────────────────────────────")
print(df_out.tail(15)[["book_title","movie_title","composite_mismatch",
                        "semantic_distance","tfidf_distance","genre_shift_jsd"
                        ]].to_string(index=False))

print("\n── TOP 15 MOST DIVERGED ─────────────────────────────────")
print(df_out.head(15)[["book_title","movie_title","composite_mismatch",
                        "semantic_distance","tfidf_distance","genre_shift_jsd"
                        ]].to_string(index=False))


# ─────────────────────────────────────────────────────────────────────────────
# CELL 10 — SANITY CHECK
# ─────────────────────────────────────────────────────────────────────────────
print(f"\n{'='*55}")
print("SANITY CHECK — Known adaptations")
print(f"{'='*55}")

known_faithful  = ["no country for old men","silence of the lambs",
                   "harry potter","jane eyre","high fidelity","big sleep",
                   "to kill a mockingbird","sense and sensibility",
                   "pride and prejudice"]
known_divergent = ["the shining","a clockwork orange","fight club",
                   "the handmaid","world war z","i am legend",
                   "apocalypse now","eyes wide shut"]

df_lower = df_out.copy()
df_lower["_bl"] = df_lower["book_title"].str.lower()

print(f"\n  Expected FAITHFUL (score should be < p67={p67:.3f}):")
for name in known_faithful:
    mask = df_lower["_bl"].str.contains(name, na=False)
    if mask.any():
        row = df_lower[mask].iloc[0]
        ok  = "" if row["composite_mismatch"] < p67 else ""
        # Diagnose unexpected direction
        head_scores = {
            "sem": row["semantic_distance"],
            "genre": row["genre_shift_jsd"],
            "tfidf": row["tfidf_distance"],
            "entity": row["entity_overlap_score"],
        }
        top_head = max(head_scores, key=head_scores.get)
        note = f"(highest head: {top_head}={head_scores[top_head]:.2f})" \
               if ok == "" else ""
        print(f"  {ok} {row['book_title']:<35} score={row['composite_mismatch']:.3f} "
              f"[{row['mismatch_category']}] {note}")
    else:
        print(f"  '{name}' — not in dataset")

print(f"\n  Expected DIVERGENT (score should be > p33={p33:.3f}):")
for name in known_divergent:
    mask = df_lower["_bl"].str.contains(name, na=False)
    if mask.any():
        row = df_lower[mask].iloc[0]
        ok  = "" if row["composite_mismatch"] > p33 else ""
        head_scores = {
            "sem": row["semantic_distance"],
            "genre": row["genre_shift_jsd"],
            "tfidf": row["tfidf_distance"],
            "entity": row["entity_overlap_score"],
        }
        top_head = max(head_scores, key=head_scores.get)
        note = f"(highest head: {top_head}={head_scores[top_head]:.2f})" \
               if ok == "" else ""
        print(f"  {ok} {row['book_title']:<35} score={row['composite_mismatch']:.3f} "
              f"[{row['mismatch_category']}] {note}")
    else:
        print(f"  '{name}' — not in dataset")

# Specific note on The Shining if present
shining_rows = df_lower[df_lower["_bl"].str.contains("shining", na=False)]
if len(shining_rows) > 0:
    row = shining_rows.iloc[0]
    print(f"\n  NOTE — The Shining analysis:")
    print(f"    Kubrick's film inverts the book's themes but shares all the same")
    print(f"    characters, locations and surface plot elements.")
    print(f"    Entity overlap ({row['entity_overlap_score']:.3f}) and TF-IDF ({row['tfidf_distance']:.3f})")
    print(f"    both pull the composite score DOWN despite genuine divergence.")
    print(f"    Semantic distance ({row['semantic_distance']:.3f}) is the most informative head here.")
    print(f"    This is a known limitation of surface-level metrics for thematic analysis.")


# ─────────────────────────────────────────────────────────────────────────────
# CELL 11 — VISUALIZATION
# ─────────────────────────────────────────────────────────────────────────────
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
import seaborn as sns

colors    = {"Faithful":"#2ecc71","Moderate":"#f39c12","Significant":"#e74c3c"}
cat_order = ["Faithful","Moderate","Significant"]

# Re-parse list columns (they become strings after CSV round-trip)
def reparse_list_col(col):
    return col.apply(lambda x: ast.literal_eval(x) if isinstance(x, str) else
                     (x if isinstance(x, list) else []))

for col in ["book_genres_list","movie_genres_list",
            "book_genres_predicted","movie_genres_predicted"]:
    df_out[col] = reparse_list_col(df_out[col])

# Strategy label for boxplot
strategy_map = {
    "pass1_wikidata_P144":       "Gold (Wikidata)",
    "pass1_wikidata_P144_batch": "Gold (Wikidata)",
    "pass2_book_adaptations":    "Silver (Book page)",
    "pass3_author_category":     "Silver (Author cat)",
    "pass4_fuzzy_verified":      "Bronze (Fuzzy)",
}
df_out["strategy_label"] = df_out["strategy"].map(strategy_map).fillna("Other")

fig = plt.figure(figsize=(20, 24))
gs  = gridspec.GridSpec(4, 2, figure=fig, hspace=0.45, wspace=0.35)

# Plot 1: Composite score distribution
ax1 = fig.add_subplot(gs[0,0])
for cat in cat_order:
    sub = df_out[df_out["mismatch_category"]==cat]["composite_mismatch"]
    ax1.hist(sub, bins=25, alpha=0.7, label=cat, color=colors[cat])
ax1.axvline(p33, color="gray", ls="--", lw=1, label=f"p33={p33:.2f}")
ax1.axvline(p67, color="gray", ls="-",  lw=1, label=f"p67={p67:.2f}")
ax1.set_title(f"Composite Mismatch Score Distribution\n"
              f"({len(df_out)} pairs after wrong-match filter)", fontweight="bold")
ax1.set_xlabel("Composite Mismatch Score")
ax1.set_ylabel("Count")
ax1.legend()

# Plot 2: Head correlation heatmap
ax2 = fig.add_subplot(gs[0,1])
corr_cols = ["composite_mismatch","semantic_distance","genre_shift_jsd",
             "tfidf_distance","entity_overlap_score","sentiment_delta"]
corr   = df_out[corr_cols].corr()
labels = ["Composite","Semantic","Genre","TF-IDF","Entity","Sentiment"]
sns.heatmap(corr, annot=True, fmt=".2f", cmap="RdYlGn_r",
            ax=ax2, square=True, linewidths=0.5,
            xticklabels=labels, yticklabels=labels)
ax2.set_title("Head Correlation Matrix\n(5 heads + composite)", fontweight="bold")

# Plot 3: Semantic vs TF-IDF scatter
ax3 = fig.add_subplot(gs[1,0])
for cat in cat_order:
    sub = df_out[df_out["mismatch_category"]==cat]
    ax3.scatter(sub["semantic_distance"], sub["tfidf_distance"],
                alpha=0.4, s=14, label=cat, color=colors[cat])
ax3.set_xlabel("Semantic Distance (embedding)")
ax3.set_ylabel("TF-IDF Distance (vocabulary)")
ax3.set_title("Semantic vs Vocabulary Distance\n(wrong matches removed)", fontweight="bold")
ax3.legend()

# Plot 4: Sentiment direction
ax4 = fig.add_subplot(gs[1,1])
ax4.hist(df_out["sentiment_direction"], bins=35,
         color="steelblue", alpha=0.8, edgecolor="white")
ax4.axvline(0, color="red", ls="--", lw=1.5, label="No change")
ax4.axvline(df_out["sentiment_direction"].mean(), color="orange", lw=1.5,
            label=f"Mean={df_out['sentiment_direction'].mean():.3f}")
ax4.set_title("Sentiment Direction\n(positive = movie lighter than book)",
              fontweight="bold")
ax4.set_xlabel("Movie Sentiment − Book Sentiment")
ax4.legend()

# Plot 5: Strategy quality boxplot
ax5 = fig.add_subplot(gs[2,0])
strat_order  = ["Gold (Wikidata)","Silver (Book page)","Silver (Author cat)","Bronze (Fuzzy)"]
strat_colors = {"Gold (Wikidata)":"#f1c40f","Silver (Book page)":"#95a5a6",
                "Silver (Author cat)":"#7f8c8d","Bronze (Fuzzy)":"#cd7f32"}
data_by_strat   = [df_out[df_out["strategy_label"]==s]["composite_mismatch"].values
                   for s in strat_order if s in df_out["strategy_label"].values]
labels_present  = [s for s in strat_order if s in df_out["strategy_label"].values]
bp = ax5.boxplot(data_by_strat, tick_labels=labels_present, patch_artist=True,
                 medianprops=dict(color="black", linewidth=2))
for patch, label in zip(bp["boxes"], labels_present):
    patch.set_facecolor(strat_colors.get(label, "#3498db"))
    patch.set_alpha(0.8)
ax5.set_title("Mismatch Score by Verification Strategy\n"
              "(Gold outliers near 0 = faithful adaptations)", fontweight="bold")
ax5.set_ylabel("Composite Mismatch Score")

# Plot 6: Genre transition heatmap  — FIXED: uses predicted genres
ax6 = fig.add_subplot(gs[2,1])
genre_transition = np.zeros((NUM_GENRES, NUM_GENRES))
for _, row in df_out.iterrows():
    for bg in row["book_genres_predicted"]:      # ← predicted (was movie_genres_list)
        if bg in TARGET_GENRES:
            bi = TARGET_GENRES.index(bg)
            for mg in row["movie_genres_predicted"]:   # ← predicted
                if mg in TARGET_GENRES:
                    mi = TARGET_GENRES.index(mg)
                    genre_transition[bi, mi] += 1

row_sums = genre_transition.sum(axis=1, keepdims=True)
genre_transition_norm = np.where(row_sums > 0, genre_transition / row_sums, 0)
sn = ["Act","Com","Dra","Hor","Thr","Sci","Fan","Rom","Ani","Doc","Lit","Non"]
sns.heatmap(genre_transition_norm, annot=True, fmt=".2f", cmap="Blues",
            ax=ax6, xticklabels=sn, yticklabels=sn, linewidths=0.3, vmin=0, vmax=0.6)
ax6.set_title("Genre Transition: P(movie genre | book genre)\n"
              "(RoBERTa-predicted — fixed from all-zeros bug)", fontweight="bold")
ax6.set_xlabel("Movie Genre")
ax6.set_ylabel("Book Genre")

# Plot 7: Top 10 most/least faithful
ax7 = fig.add_subplot(gs[3,:])
n        = 10
top_div  = df_out.nlargest(n, "composite_mismatch")
top_fai  = df_out.nsmallest(n, "composite_mismatch")
examples = pd.concat([top_div, top_fai])
labels_bar = [f"{r['book_title'][:24]} → {r['movie_title'][:24]}"
              for _, r in examples.iterrows()]
scores_bar = examples["composite_mismatch"].values
bar_colors = ["#e74c3c"]*n + ["#2ecc71"]*n
ax7.barh(range(len(labels_bar)), scores_bar, color=bar_colors, alpha=0.85)
ax7.set_yticks(range(len(labels_bar)))
ax7.set_yticklabels(labels_bar, fontsize=8)
ax7.axvline(p33, color="gray", ls="--", lw=1)
ax7.axvline(p67, color="gray", ls="-",  lw=1)
ax7.set_title("Top 10 Most Diverged (red) & Most Faithful (green)\n"
              "(wrong-match filtered — these are genuine adaptation differences)",
              fontweight="bold")
ax7.set_xlabel("Composite Mismatch Score")
ax7.invert_yaxis()

plt.suptitle(
    f"Book-to-Movie Mismatch Analysis\n"
    f"{len(df_out)} pairs | Wikipedia plots | 5-head scoring | wrong-match filtered",
    fontsize=14, fontweight="bold", y=1.01
)

plt.savefig(f"{SAVE}wiki_mismatch_analysis.png", dpi=150, bbox_inches="tight")
plt.savefig(f"{DATA}wiki_mismatch_analysis.png", dpi=150, bbox_inches="tight")
plt.show()
print("Plots saved.")
print(f"\n Pipeline complete.")
print(f"   wiki_mismatch_scores.csv  — {len(df_out)} rows")
print(f"   wiki_wrong_matches.csv    — {len(df_wrong)} rows (scraper errors, excluded)")
print(f"   wiki_mismatch_analysis.png")
```
